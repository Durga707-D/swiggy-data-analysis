```python

df['Order_Date'] = pd.to_datetime(df['Order_Date'])

df['Order_Day'] = df['Order_Date'].dt.day_name()

print(df[['Order_Date', 'Order_Day']].head())
```

```python

df['Order_Hour'] = pd.to_numeric(
    df['Order_Time'].astype(str).str[:2],
    errors='coerce'
)

print(df[['Order_Time', 'Order_Hour']].head(10))
```

### Recommendations and Improvements

Based on the analysis of Swiggy food delivery data, the following improvements can be suggested:

1. **Improve Delivery Services:** Swiggy should focus on cities with higher average delivery times and improve delivery partner availability in those areas.

2. **Focus on Popular Food Categories:** Restaurants should maintain sufficient ingredients and staff for food categories that receive more orders.

3. **Improve Customer Satisfaction:** Since delivery time has a negative relationship with customer ratings in our analysis, reducing unnecessary delivery delays may help improve customer experience.

4. **Plan for Peak Hours:** More delivery partners should be available during busy ordering hours to handle increased demand.

5. **Improve Restaurant Performance:** Restaurants receiving lower ratings should work on food quality, packaging and preparation time.

6. **Use Customer Preferences:** Payment methods, popular restaurants and ordering trends can help Swiggy plan better offers and services.

7. **Support Business Growth:** Swiggy can use city-wise and category-wise order patterns to plan marketing campaigns and improve restaurant partnerships.

## Key Findings and Insights

The analysis of Swiggy food delivery data helped us understand customer ordering behaviour, restaurant performance and delivery patterns.

1. **Customer Ordering Patterns:** Order values, popular food categories, restaurants and cities helped us understand customer preferences.

2. **Delivery Performance:** The correlation between order value and delivery time was approximately 0.001, showing almost no linear relationship between these two variables.

3. **Customer Satisfaction:** The correlation between delivery time and customer ratings was approximately -0.389, indicating that longer delivery times tend to be associated with lower ratings.

4. **City-wise Performance:** Comparing average order values, delivery times and ratings helps identify differences in business performance across cities.

5. **Ordering Trends:** Analysing orders by day, hour and food category helps identify busy periods and customer preferences at different times.

6. **Business Improvement:** The results can help Swiggy plan delivery partners, improve restaurant services and manage demand during busy hours.

# Swiggy Food Delivery Dashboard

This dashboard provides a visual overview of customer orders, restaurant performance, delivery trends and customer satisfaction using Python.

```python

total_orders = df['Order_ID'].nunique()
average_order_value = df['Order_Value'].mean()
average_rating = df['Rating'].mean()
average_delivery_time = df['Delivery_Time'].mean()

print("Total Orders:", total_orders)
print("Average Order Value:", round(average_order_value, 2))
print("Average Customer Rating:", round(average_rating, 2))
print("Average Delivery Time:", round(average_delivery_time, 2), "minutes")
```

```python

fig, ax = plt.subplots(1, 4, figsize=(16, 4))

kpi_titles = [
    'Total Orders',
    'Average Order Value',
    'Average Rating',
    'Average Delivery Time'
]

kpi_values = [
    f'{total_orders:,}',
    f'₹{average_order_value:.2f}',
    f'{average_rating:.2f}/5',
    f'{average_delivery_time:.1f} min'
]

for i in range(4):
    ax[i].text(
        0.5, 0.55, kpi_values[i],
        ha='center', va='center',
        fontsize=18, fontweight='bold'
    )

    ax[i].text(
        0.5, 0.25, kpi_titles[i],
        ha='center', va='center',
        fontsize=10
    )

    ax[i].axis('off')

plt.suptitle('SWIGGY FOOD DELIVERY — KEY PERFORMANCE INDICATORS',
             fontsize=15, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python

city_orders = df['City'].value_counts().sort_values()

plt.figure(figsize=(10, 6))

city_orders.plot(
    kind='barh',
    color='coral',
    edgecolor='black'
)

plt.title('Total Orders by City', fontsize=14, fontweight='bold')
plt.xlabel('Number of Orders')
plt.ylabel('City')
plt.grid(axis='x', linestyle='--', alpha=0.4)
plt.tight_layout()

plt.show()
```

```python

category_orders = df['Food_Category'].value_counts().sort_values()

plt.figure(figsize=(10, 6))

category_orders.plot(
    kind='barh',
    color='skyblue',
    edgecolor='black'
)

plt.title('Most Popular Food Categories', fontsize=14, fontweight='bold')
plt.xlabel('Number of Orders')
plt.ylabel('Food Category')
plt.grid(axis='x', linestyle='--', alpha=0.4)
plt.tight_layout()

plt.show()
```

```python

weekday_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

day_orders = df['Order_Day'].value_counts().reindex(weekday_order)

plt.figure(figsize=(10, 5))

day_orders.plot(
    kind='bar',
    color='orange',
    edgecolor='black'
)

plt.title('Orders by Day of the Week', fontsize=14, fontweight='bold')
plt.xlabel('Day')
plt.ylabel('Number of Orders')
plt.xticks(rotation=30)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()

plt.show()
```

```python

hour_orders = df['Order_Hour'].value_counts().sort_index()

plt.figure(figsize=(12, 5))

hour_orders.plot(
    kind='bar',
    color='mediumseagreen',
    edgecolor='black'
)

plt.title('Orders by Hour of the Day', fontsize=14, fontweight='bold')
plt.xlabel('Hour (24-hour format)')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()

plt.show()
```

```python

import matplotlib.pyplot as plt

fig = plt.figure(figsize=(15, 9), facecolor='#F7F8FA')

gs = fig.add_gridspec(
    3, 4,
    height_ratios=[0.9, 2, 2],
    hspace=0.55,
    wspace=0.45
)

fig.suptitle(
    'SWIGGY FOOD DELIVERY ANALYSIS',
    fontsize=19,
    fontweight='bold',
    color='#262626',
    y=0.98
)

# 1. KPI Cards
kpis = [
    ('TOTAL ORDERS', f'{total_orders:,}'),
    ('AVG ORDER VALUE', f'₹{average_order_value:.2f}'),
    ('AVG RATING', f'{average_rating:.2f}/5'),
    ('AVG DELIVERY TIME', f'{average_delivery_time:.1f} min')
]

for i, (title, value) in enumerate(kpis):
    ax = fig.add_subplot(gs[0, i])
    ax.set_facecolor('#E8F0FE')

    ax.text(0.5, 0.65, value,
            ha='center', va='center',
            fontsize=17, fontweight='bold',
            color='#243B53')

    ax.text(0.5, 0.25, title,
            ha='center', va='center',
            fontsize=9, fontweight='bold',
            color='#52616B')

    ax.set_xticks([])
    ax.set_yticks([])

    for spine in ax.spines.values():
        spine.set_visible(False)

# 2. Orders by City
ax1 = fig.add_subplot(gs[1, 0:2])
city_orders.sort_values().plot(
    kind='barh', ax=ax1, color='#F28E2B'
)
ax1.set_title('Orders by City', fontweight='bold')
ax1.set_xlabel('Number of Orders')
ax1.set_ylabel('')
ax1.grid(axis='x', alpha=0.2)

# 3. Popular Food Categories
ax2 = fig.add_subplot(gs[1, 2:4])
category_orders.sort_values().plot(
    kind='barh', ax=ax2, color='#4E9F9D'
)
ax2.set_title('Popular Food Categories', fontweight='bold')
ax2.set_xlabel('Number of Orders')
ax2.set_ylabel('')
ax2.grid(axis='x', alpha=0.2)

# 4. Orders by Day
ax3 = fig.add_subplot(gs[2, 0:2])
day_orders.plot(
    kind='bar', ax=ax3, color='#7B83EB'
)
ax3.set_title('Orders by Day', fontweight='bold')
ax3.set_xlabel('')
ax3.set_ylabel('Number of Orders')
ax3.tick_params(axis='x', rotation=25)
ax3.grid(axis='y', alpha=0.2)

# 5. Orders by Hour
ax4 = fig.add_subplot(gs[2, 2:4])
hour_orders.plot(
    kind='bar', ax=ax4, color='#E76F8A'
)
ax4.set_title('Orders by Hour', fontweight='bold')
ax4.set_xlabel('Hour of the Day')
ax4.set_ylabel('Number of Orders')
ax4.tick_params(axis='x', rotation=0)
ax4.grid(axis='y', alpha=0.2)

plt.savefig(
    'Swiggy_Dashboard.png',
    dpi=300,
    bbox_inches='tight',
    facecolor=fig.get_facecolor()
)

plt.show()
```
